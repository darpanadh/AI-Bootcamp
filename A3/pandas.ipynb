Creating table for the data

```python
import pandas as pd

data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],
    'Age': [20, 22, 19, 21, 23],
    'Score': [85, 92, 78, 95, 88],
    'Passed': [True, True, False, True, True]
}

# Turn the dictionary into a Pandas DataFrame (Table)
df = pd.DataFrame(data)

print(df)
```

Data inspection

```python
# 1. Look at the top 3 rows
print("--- First 3 Rows ---")
print(df.head(3))

# 2. Check table summary (column types, non-null counts)
print("\n--- Info ---")
df.info()

# 3. Quick statistical summary of numerical columns
print("\n--- Summary Statistics ---")
print(df.describe())
```

Selecting column and row

```python
# Select a single column
scores = df['Score']
print("Scores Column:\n", scores)

# Select multiple columns =
subset = df[['Student', 'Score']]
print("\nStudent & Score Only:\n", subset)
```

Selecting Rows by Position (.iloc):

```python
# Get the first row (Row 0)
first_row = df.iloc[0]
print("First Row:\n", first_row)

# Get rows 0 up to 2 (Rows 0 and 1)
first_two_rows = df.iloc[0:2]
print("\nFirst Two Rows:\n", first_two_rows)
```

Basic Aggregation

```python
# Overall average score
print("Average Score:", df['Score'].mean())

# Highest score
print("Max Score:", df['Score'].max())

# Count how many students passed vs failed
print("\nPass/Fail Counts:")
print(df['Passed'].value_counts())
```
