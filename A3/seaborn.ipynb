```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

```python
tips = sns.load_dataset('tips')

# View the first few rows
print(tips.head())
```

Scatter plot

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time')

plt.title("Total Bill vs Tip Amount")
plt.show()
```

Categorical Bar Plots

```python
# Calculate and plot the average total bill per day
sns.barplot(data=tips, x='day', y='total_bill', color='Blue')

plt.title("Average Total Bill by Day")
plt.show()
```

Histogram

```python
# Plot distribution of bill amounts with a smooth density curve (kde)
sns.histplot(data=tips, x='total_bill', kde=True, color='purple')

plt.title("Distribution of Total Bills")
plt.show()
```

Box Plots

```python
# Compare bill distributions across days
sns.boxplot(data=tips, x='day', y='total_bill', color='green')

plt.title("Bill Range and Outliers by Day")
plt.show()
```
