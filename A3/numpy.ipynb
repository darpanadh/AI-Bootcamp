Basic array operation 

```python
import numpy as np
marks = np.array([86,54,89,98,94])
print("Marks:",marks)

print("Average score:",np.mean(marks))
print("Highest score:",np.max(marks))
print("Lowest score:",np.min(marks))
print("Sum of scores:",np.sum(marks))
print("Standard deviation:",np.std(marks))
```

Checking attributes

```python
print("Data type:", marks.dtype)
print("Shape (Dimensions):", marks.shape)
print("Total elements:", marks.size)
```

```python
# Create an array of sequential numbers from 1 to 10
range_array = np.arange(1, 11)
print("Range 1-10:", range_array)

# Create an array filled with zeros or ones
zeros = np.zeros(5)
ones = np.ones(5)
print("Zeros:", zeros)
print("Ones:", ones)
```

```python
# Access elements by position (0-indexed)
print("First mark:", marks[0])
print("Last mark:", marks[-1])

# Slice elements (get index 1 up to index 3)
print("Middle marks:", marks[1:4])
```

```python
# A list of daily temperature readings in Celsius
temps = np.array([18, 22, 15, 30, 27, 14, 31])

# Find all temperatures above 25 degrees
hot_days = temps[temps > 25]

print("All Temperatures:", temps)
print("Hot Days (> 25°C):", hot_days)
```
