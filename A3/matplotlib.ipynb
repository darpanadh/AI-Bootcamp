Line Plots

```python
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
study_hours = [2, 4, 3, 5, 1]

# Step 2: Create Line Plot with Styling
plt.plot(days, study_hours, color='green', marker='o', linestyle='--')

# Add Labels and Title
plt.title("Study Hours Throughout the Week")
plt.xlabel("Day of the Week")
plt.ylabel("Hours Spent Studying")

# Step 3: Display Plot
plt.show()
```

Bar Chart

```python
fruits = ['Apples', 'Bananas', 'Oranges', 'Grapes']
quantity = [25, 40, 15, 30]

plt.bar(fruits, quantity, color='green')

plt.title("Fruits Market Inventory")
plt.xlabel("Fruits")
plt.ylabel("Quantity")

plt.show()
```

Scatter Plot

```python
temperatures = [20, 25, 28, 32, 35, 40]
sales_in_usd = [120, 200, 260, 310, 400, 520]

# s = dot size
plt.scatter(temperatures, sales_in_usd, color="blue", s=100)

plt.title("Ice Cream Sales vs Temperature")
plt.xlabel("Temperature (°C)")
plt.ylabel("Sales ($)")

plt.grid()
plt.show()
```

Multiple plot

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
student_A = [2, 4, 3, 5, 1]
student_B = [3, 3, 4, 2, 5]

# Layer two lines
plt.plot(days, student_A, label='Student A', color='blue', marker='o')
plt.plot(days, student_B, label='Student B', color='orange', marker='s')

# Add Titles and Legend
plt.title("Study Hours Comparison")
plt.xlabel("Day")
plt.ylabel("Hours")
plt.legend() # Displays the box showing which color belongs to which student

plt.show()
```
